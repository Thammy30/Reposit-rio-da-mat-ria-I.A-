21. Busca Binária

 * Implemente a busca binária para encontrar um elemento em uma lista ordenada.

 EXPLICAÇÃO:

 Busca Binária: Algoritmo clássico que divide o problema ao meio, sendo muito mais rápido que buscas lineares.

```python
numeros = [5, 7, 9, 35, 42, 55, 60, 78]

numero = int(input("Digite um número: "))

inicio = 0
fim = len(numeros) - 1

while inicio <= fim:
    meio = (inicio + fim) // 2

    if numeros[meio] == numero:
        print(f"Número encontrado na posição {meio}.")
        break

    elif numero > numeros[meio]:
        inicio = meio + 1
    else:
        fim = meio - 1
else:
    print("Número não encontrado na lista")


```

22. Criptografia de Cifra de César

 * Crie uma função para criptografar uma mensagem usando a cifra de César

 EXPLICAÇÃO:

 A Cifra de César é uma técnica simples de criptografia que substitui cada letra de uma mensagem por outra letra do alfabeto, seguindo um número determinado de posições.

```python
def criptografar_cesar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    resultado = ""

    for letra in mensagem.lower():
        if letra in alfabeto:
            posicao = alfabeto.find(letra)
            nova_posicao = (posicao + deslocamento) % 26
            resultado += alfabeto[nova_posicao]
        else:
            resultado += letra

    return resultado


texto = input("Digite a mensagem: ")
chave = int(input("Digite a chave de deslocamento (ex: 3): "))

mensagem_cifrada = criptografar_cesar(texto, chave)

print(f"Mensagem criptografada: {mensagem_cifrada}")
```

23. Descriptografar Cifra de César

 * Crie uma função para descriptografar uma mensagem cifrada com a cifra de César

 EXPLICAÇÃO:

 Cifra de César: A descriptografia faz o processo contrário: pega a mensagem cifrada e volta o número de posições definido pela chave para recuperar a mensagem original.

```python
def descriptografar_cesar(mensagem, deslocamento):
    alfabeto = "abcdefghijklmnopqrstuvwxyz"
    resultado = ""

    for letra in mensagem.lower():
        if letra in alfabeto:
            posicao = alfabeto.find(letra)

            nova_posicao = (posicao - deslocamento) % 26

            resultado += alfabeto[nova_posicao]
        else:
            resultado += letra

    return resultado


texto_cifrado = input("Digite a mensagem para decifrar: ")
chave = int(input("Digite a chave (o número usado para cifrar): "))

texto_revelado = descriptografar_cesar(texto_cifrado, chave)
print(f"Mensagem original: {texto_revelado}")

```

24. Validação de CPF

 * Implemente uma função para validar um CPF brasileiro.

 EXPLICAÇÃO:

 Validação de CPF: Verifica se o CPF possui 11 dígitos válidos e utiliza cálculos matemáticos para confirmar os dois dígitos verificadores. Também identifica formatos inválidos, como sequências repetidas.

```python
def validar_cpf(cpf):
    cpf = cpf.replace(".", "").replace("-", "")

    if len(cpf) != 11:
        return False

    if not cpf.isdigit():
        return False

    if cpf == cpf[0] * 11:
        return False

    return True


cpf = input("Digite seu CPF: ")

if validar_cpf(cpf):
    print("CPF válido")
else:
    print("CPF inválido")
```

30. Jogo da Forca

 * Desenvolva um jogo simples da forca onde o usuário deve adivinhar uma palavra.

 EXPLICAÇÃO:

 Forca: Gerenciamento de strings e controle de tentativas.

```python
import random

palavras = ["python", "dados", "estatistica", "algoritmo", "biblioteca"]

palavra_secreta = random.choice(palavras)
letras_descobertas = ["_" for _ in palavra_secreta]

tentativas = 10
letras_tentadas = []

print("--- BEM-VINDO AO JOGO DA FORCA ---")

while tentativas > 0 and "_" in letras_descobertas:
    print(f"\nPalavra: {' '.join(letras_descobertas)}")
    print(f"Tentativas restantes: {tentativas}")
    print(f"Letras já tentadas: {', '.join(letras_tentadas)}")

    chute = input("Digite uma letra: ").lower()

    if chute in letras_tentadas:
        print("Você já tentou essa letra. Tente outra.")
        continue

    letras_tentadas.append(chute)

    if chute in palavra_secreta:
        print("Boa! A letra existe na palavra.")

        for i, letra in enumerate(palavra_secreta):
            if letra == chute:
                letras_descobertas[i] = chute

    else:
        print("Que pena, essa letra não existe.")
        tentativas -= 1


if "_" not in letras_descobertas:
    print(f"\nParabéns! Você venceu. A palavra era: {palavra_secreta}")
else:
    print(f"\nGame Over! Você perdeu. A palavra era: {palavra_secreta}")
```

31. Ordenação por Bolha

 * Implemente o algoritmo de ordenação por bolha (Bubble Sort).

 EXPLICAÇÃO:

 Bubble Sort: Ensina como algoritmos de ordenação movem elementos na memória.

```python
def bubble_sort(lista):
    n = len(lista)

    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

numeros = [64, 34, 25, 12, 22, 11, 90]
print(f"Lista original: {numeros}")

lista_ordenada = bubble_sort(numeros)
print(f"\nLista ordenada: {lista_ordenada}")
```

 EXPLICAÇÃO:

 Jogo da Velha: Uso de matrizes 3x3 e verificação de condições de vitória (linhas, colunas, diagonais).

```python
tabuleiro = [[" " for _ in range(3)] for _ in range(3)]


def imprimir_tabuleiro(tab):
    for linha in tab:
        print(f" {linha[0]} | {linha[1]} | {linha[2]} ")
        print("-----------")


def verificar_vitoria(tab, jogador):

    for linha in tab:
        if linha[0] == jogador and linha[1] == jogador and linha[2] == jogador:
            return True


    for coluna in range(3):
        if tab[0][coluna] == jogador and tab[1][coluna] == jogador and tab[2][coluna] == jogador:
            return True


    if tab[0][0] == jogador and tab[1][1] == jogador and tab[2][2] == jogador:
        return True


    if tab[0][2] == jogador and tab[1][1] == jogador and tab[2][0] == jogador:
        return True

    return False


def jogo_da_velha():
    tabuleiro = [[" " for _ in range(3)] for _ in range(3)]
    jogador_atual = "X"

    for turno in range(9):
        imprimir_tabuleiro(tabuleiro)

        print(f"Turno do jogador {jogador_atual}")

        linha = int(input("Escolha a linha (0, 1, 2): "))
        coluna = int(input("Escolha a coluna (0, 1, 2): "))

        if tabuleiro[linha][coluna] != " ":
            print("Posição já ocupada! Tente novamente.")
            continue

        tabuleiro[linha][coluna] = jogador_atual


        if verificar_vitoria(tabuleiro, jogador_atual):
            imprimir_tabuleiro(tabuleiro)
            print(f"\n🎉 Jogador {jogador_atual} venceu!")
            break


        jogador_atual = "O" if jogador_atual == "X" else "X"

    else:
        imprimir_tabuleiro(tabuleiro)
        print("\nDeu velha! Ninguém venceu.")


jogo_da_velha()
```

35. Números Fibonacci Até N

 * Gere a sequência de Fibonacci até um número n fornecido pelo usuário.

 EXPLICAÇÃO:

 Fibonacci até N: Loop while com condição de parada baseada no valor final.

```python
def fibonacci_ate_n(n):
    sequencia = []
    a, b = 0, 1


    while a <= n:
        sequencia.append(a)


        a, b = b, a + b

    return sequencia


try:
    limite = int(input("Gerar Fibonacci até qual número? "))
    resultado = fibonacci_ate_n(limite)
    print(f"Sequência: {resultado}")
except ValueError:
    print("Por favor, digite um número inteiro válido.")
```

36. Gerador de Senhas

 * Crie um gerador de senhas aleatóriascom letras e números.

 EXPLICAÇÃO:

 Gerador de Senhas: Sorteio aleatório com random.choice() e união de caracteres com "".join()

```python
import random

def gerar_senha(tamanho):
    caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"

    senha_lista = [random.choice(caracteres) for _ in range(tamanho)]

    senha_final = "".join(senha_lista)

    return senha_final



try:
    comprimento = int(input("Digite o tamanho da senha desejada: "))
    nova_senha = gerar_senha(comprimento)

    print(f"Sua senha gerada é: {nova_senha}")

except ValueError:
    print("Por favor, digite um número inteiro para o tamanho.")
```

37.  Funções Recursivas

 EXPLICAÇÃO:

 Recursividade: Introduz funções que chamam a si mesmas, essencial para algoritmos de Árvores de Decisão.

```python
def calcular_fatorial(n):

    if n <= 1:
        return 1


    else:

        return n * calcular_fatorial(n - 1)


try:
    numero = int(input("Digite um número para calcular o fatorial: "))
    if numero < 0:
        print("Fatorial não existe para números negativos.")
    else:
        resultado = calcular_fatorial(numero)
        print(f"O fatorial de {numero} é {resultado}")
except ValueError:
    print("Por favor, digite um número inteiro válido.")
```

40. Lista de Compras

 * Crie um programa para gerenciar uma lista de compras, permitindo adicionar, remover e listar itens.

 EXPLICAÇÃO:

 Lista de Compras: Permite adicionar, remover e visualizar produtos de forma simples, utilizando comandos como append() e remove() para organizar a lista.


```python
def menu_compras():
    lista_compras = []

    while True:
        print("\n--- LISTA DE COMPRAS ---")
        print("1. Adicionar item")
        print("2. Remover item")
        print("3. Listar itens")
        print("4. Sair")

        opcao = input("Escolha uma opção: ")

        if opcao == '1':
            item = input("Digite o nome do produto: ")
            lista_compras.append(item)
            print(f"'{item}' adicionado!")

        elif opcao == '2':
            item = input("Qual item deseja remover? ")
            if item in lista_compras:
                lista_compras.remove(item)
                print(f"'{item}' removido com sucesso.")
            else:
                print("Erro: Item não encontrado na lista.")

        elif opcao == '3':
            print("\nSUA LISTA ATUAL:")
            if not lista_compras:
                print("A lista está vazia.")
            else:

                for i, item in enumerate(lista_compras, 1):
                    print(f"{i}. {item}")

        elif opcao == '4':
            print("Programa encerrado.")
            break
        else:
            print("Opção inválida.")


menu_compras()
```

42. Números Primos até N

 * Gere todos os números primos até um número n fornecido.

 EXPLICAÇÃO:

 Primos até N: Verifica os números de um intervalo e identifica quais são primos, utilizando repetição e condições para testar seus divisores.

```python
def eh_primo(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True


def gerar_primos_ate_n(n):
    primos = []

    for x in range(2, n + 1):
        if eh_primo(x):
            primos.append(x)

    return primos


limite = int(input("Gerar primos até qual número? "))

resultado = gerar_primos_ate_n(limite)

print(f"Números primos encontrados: {resultado}")
```

43. Conversor de Unidades

 * Converta unidades de comprimento (por exemplo, metros para centímetros).

 EXPLICAÇÃO:

 Conversor de Unidades: Uso de dicionários como mapas de conversão para evitar múltiplos ifs.

```python
def conversor_comprimento(valor, unidade_origem, unidade_destino):
    fatores = {
        "m": 1,
        "cm": 100,
        "mm": 1000,
        "km": 0.001
    }

    if unidade_origem not in fatores or unidade_destino not in fatores:
        return None

    valor_em_metros = valor / fatores[unidade_origem]
    resultado = valor_em_metros * fatores[unidade_destino]

    return resultado


num = float(input("Digite o valor numérico: "))
de = input("Unidade de origem (m, cm, mm, km): ").lower()
para = input("Unidade de destino (m, cm, mm, km): ").lower()

res = conversor_comprimento(num, de, para)

if res is not None:
    print(f"{num}{de} é igual a {res:.4f}{para}")
else:
    print("Unidade inválida.")
```

44. Fibonacci em Nível

 * Gere a sequência de Fibonacci até um determinado número de termos.

 EXPLICAÇÃO:

 Fibonacci Termos: Controle de loop baseado no tamanho da lista (len)


```python
def fibonacci_por_termos(n_termos):
    sequencia = []
    a, b = 0, 1

    if n_termos <= 0:
        return []

    while len(sequencia) < n_termos:
        sequencia.append(a)

        a, b = b, a + b

    return sequencia


try:
    quantidade = int(input("Quantos termos da sequência você deseja? "))

    resultado = fibonacci_por_termos(quantidade)

    print(f"\nSequência com {quantidade} termos: {resultado}")

except ValueError:
    print("Erro: Digite um número inteiro para a quantidade.")
```

46. Calcular Média e Desvio Padrão.

 * Calcule a média e o desvio padrão de uma lista de números.

EXPLICAÇÃO:

Média e Desvio Padrão: Cálculo da variabilidade dos dados. O desvio padrão usa n−1 para ser uma estimativa não enviesada


```python
import math

def calcular_estatisticas(lista):
    n = len(lista)

    if n < 2:
        return "Erro: A lista precisa de pelo menos dois números."

    media = sum(lista) / n

    soma_quadrados_desvios = sum((x - media) ** 2 for x in lista)

    variancia = soma_quadrados_desvios / (n - 1)

    desvio_padrao = math.sqrt(variancia)

    return media, desvio_padrao


numeros = [4, 5, 6, 7, 8]

media_res, desvio_res = calcular_estatisticas(numeros)

print(f"Lista: {numeros}")
print(f"Média: {media_res}")
print(f"Desvio Padrão: {desvio_res:.2f}")
```

47. Contar Números em Lista

 * Conte quantas vezes cada número aparece em uma lista.

 EXPLICAÇÃO:

 Contar Números: Verifica quantas vezes cada número aparece na lista, utilizando um dicionário para armazenar e atualizar as quantidades.

```python
def contar_numeros_manual(lista):
    contagem = {}

    for x in lista:
        if x in contagem:
            contagem[x] += 1
        else:
            contagem[x] = 1

    return contagem


numeros = [1, 2, 2, 3, 3, 3, 4]

resultado = contar_numeros_manual(numeros)

print(f"Números: {numeros}")
print(f"\nQuantidade de cada número: {resultado}")
```

48. Ordenação por Inserção

 * Implemente o algoritmo de ordenação por inserção (Insertion Stort).


EXPLICAÇÃO:

Insertion Sort: Demonstração de como inserir dados mantendo a ordem (importante para fluxos contínuos).

```python
def insertion_sort_explicado(lista):
    print(f"Lista inicial: {lista}\n" + "-"*30)

    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1
        while j >= 0 and chave < lista[j]:
            lista[j + 1] = lista[j]
            j -= 1
        lista[j + 1] = chave

numeros = [300, 312, 500, 269, 150, 1, 2 ]
insertion_sort_explicado(numeros)
```

50. Cálculo de Frequência de Letras

 * Calcule a frequência de cada letra em uma string fornecida pelo usuário.

Explicação:

INPUT (A Entrada): O que o usuario digitar fica guardado na memória com o nome de texto_usuario.

FREQUÊNCIA {} (O Estoque): Essas chaves cria um dicionario. Ele funciona como uma tabela de contagem que começa totalmente limpa, pronta para receber as letras e suas quantidades.

FOR/IF/ELSE (O coração do codigo): O for faz o computador letra por letra, o if faz a contagem de +1 quando houver repetição e o else adicione pela primeira vez a letra na tabela valendo 1.

PRINT FREQUÊNCIA (O resultado): Por fim, o print mostra na tela a nossa tabela preenchida com o total de cada letra.

```python
def frequencia_letras_manual(texto):
    contagem = {}
    for caractere in texto.lower():

        if caractere.isalpha():
            if caractere in contagem:
                contagem[caractere] += 1
            else:
                contagem[caractere] = 1
    return contagem

# Teste
entrada = input("Digite uma frase: ")
print(frequencia_letras_manual(entrada))
```
